# Business Insights & Recommendations

## Key Insights

* Approximately **60% of orders have a missing "store_id" **, indicating that a large portion of transactions cannot currently be attributed to a specific store.
* Approximately **65% of orders have a missing "campaign_id"**, limiting the ability to accurately measure campaign-level performance.
* All orders can be linked to valid customer records, supporting reliable customer-level analysis.
* The "gross_amount" distribution contains high-value outliers that should be investigated to determine whether they represent genuine high-value transactions or potential data-quality issues.

## Recommendations

* Improve store and sales-channel tracking to ensure transactions can be attributed to the appropriate source.
* Improve campaign attribution so marketing performance and campaign effectiveness can be measured more accurately.
* Continue validating customer and order relationships as part of the data-quality process.
* Investigate high-value transactions separately before removing them as outliers, since they may represent genuine large purchases.

## Conclusion

The analysis demonstrates a structured approach to data cleaning, validation, exploratory analysis, and outlier detection. Improving data completeness and attribution would enable more reliable business and marketing analysis.
